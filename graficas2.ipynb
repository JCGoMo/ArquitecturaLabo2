```python
import numpy as np
import matplotlib.pyplot as plt
```

## Analisis de optimización de numCycles

```python
x = 1
width = 0.25  # Ancho de las barras

y_in_order = np.array([221601.0])
y_in_order_op = np.array([178256.0])
y_out_of_order_op = np.array([37560.0])

plt.figure(figsize=(10, 7))

plt.bar(x - width, y_in_order, width, label='in_order sin optimizar')
plt.bar(x, y_in_order_op, width, label='in_order optimizado')
plt.bar(x + width, y_out_of_order_op, width, label='out_of_order optimizado') 

# Configurar título y etiquetas
plt.title('Analisis de optimización de numCycles')
plt.ylabel('numCycles')
plt.xlabel('Tamaño de caché de 32kB')
plt.xticks([])

plt.legend(fontsize='large')
# Mostrar el gráfico
plt.show()
```
