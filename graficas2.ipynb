```python
import numpy as np
import matplotlib.pyplot as plt
```

# numCycles

```python
x = np.array(['8kB', '16kB', '32kB', '64kB', '128kB'])
y_1Via = np.array([221813.0, 221865.0, 221601.0, 201188.0, 246391.0])
y_2Via = np.array([187414.0, 187462.0, 187462.0, 187503.0, 246391.0])
y_4Via = np.array([246345.0, 246409.0, 246409.0, 246409.0, 246391.0])
y_8Via = np.array([246377.0, 246409.0, 246409.0, 246409.0, 246391.0])

# Crear subplots
fig, axs = plt.subplots(4, 1, figsize=(12, 16))

axs[0].set_title('Análisis de numCycles')
axs[0].plot(x, y_1Via, color='tab:purple', marker='o')
axs[0].legend(loc='upper left', title='1 vía')

axs[1].plot(x, y_2Via, color='tab:blue', marker='o')
axs[1].legend(loc='upper left', title='2 vías')

axs[2].plot(x, y_4Via, color='tab:cyan', marker='o')
axs[2].legend(loc='upper left', title='4 vías')

axs[3].plot(x, y_8Via, color='tab:green', marker='o')
axs[3].legend(loc='upper left', title='8 vías')
axs[3].set_xlabel("Número de vías")


# Ajustes de diseño para evitar solapamiento
plt.tight_layout()

# Mostrar el gráfico
plt.show()
```
