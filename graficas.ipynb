```python
import numpy as np
import matplotlib.pyplot as plt
```

#  Análisis con caché de 1 vía 

```python
# Datos
x = np.array(['8kB', '16kB', '32kB', '64kB', '128kB'])
y_numCycles = np.array([221813.0, 221865.0, 221601.0, 201188.0, 246391.0])
y_idleCycles = np.array([132471.0, 132523.0, 132250.0, 137567.0, 186794.0])
y_overallHits = np.array([802.0, 802.0, 806.0, 6114.0, 7691.0])
y_ReadReq_hits = np.array([482.0, 482.0, 480.0, 5923.0, 7691.0])

# Crear subplots
fig, axs = plt.subplots(4, 1, figsize=(15, 16))

# numCycles
axs[0].plot(x, y_numCycles, color='tab:purple', label='numCycles')
axs[0].set_title('numCycles')

# idleCycles
axs[1].plot(x, y_idleCycles, color='tab:blue', label='idleCycles')
axs[1].set_title('idleCycles')

# overallHits
axs[2].plot(x, y_overallHits, color='tab:cyan', label='overallHits')
axs[2].set_title('overallHits')

# ReadReq.hits
axs[3].plot(x, y_ReadReq_hits, color='tab:green', label='ReadReq.hits')
axs[3].set_title('ReadReq.hits')
axs[3].set_xlabel("Tamaño de caché")


# Ajustes de diseño para evitar solapamiento
plt.tight_layout()

# Mostrar el gráfico
plt.show()
```
